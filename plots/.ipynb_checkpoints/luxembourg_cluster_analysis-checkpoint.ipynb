```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import pairwise_distances_argmin_min
import networkx as nx
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering
import seaborn as sns
import pandas as pd
import math
import random
```

```python
# Definir a semente
seed_value = 20

np.random.seed(seed_value)
random.seed(seed_value-2)
```

```python
def plota_topologia(data,topology):
    # Criar um grafo direcionado
    G = nx.DiGraph()

    # Adicionar nós ao grafo com suas posições
    for node, pos in data.items():
        G.add_node(node, pos=pos)

    # Adicionar arestas ao grafo
    G.add_edges_from(topology)

    # Obter posições dos nós
    node_positions = {node: pos for node, pos in data.items()}

    # Desenhar o grafo
    plt.figure(figsize=(10, 6))
    nx.draw(
        G,
        pos=node_positions,
        node_size=100,
        node_color='lightblue',
        edge_color='gray',
        width=0.5,
        with_labels=True,
        font_size=8,
        font_color='black'
    )
    plt.title('Topologia da Rede')
    plt.show()
```

```python
# Função para reclusterizar um cluster
def recluster(data_array, cluster_indices):
    subdata = data_array[cluster_indices]
    subkmeans = KMeans(n_clusters=2, n_init=10, random_state=seed_value)
    subclusters = subkmeans.fit_predict(subdata)
    return subclusters
```

```python

data_array = np.array(list(data.values()))

# Número máximo de elementos por cluster
max_elements_per_cluster = 4

cluster_list = [list(data.keys())]

# Enquanto houver algum cluster com mais de max_elements_per_cluster elementos
while any(len(cluster) > max_elements_per_cluster for cluster in cluster_list):
    new_cluster_list = []
    for cluster_indices in cluster_list:
        if len(cluster_indices) > max_elements_per_cluster:
            subclusters = recluster(data_array, cluster_indices)
            for subcluster_id in np.unique(subclusters):
                new_cluster_indices = [cluster_indices[i] for i in range(len(cluster_indices)) if subclusters[i] == subcluster_id]
                new_cluster_list.append(new_cluster_indices)
        else:
            new_cluster_list.append(cluster_indices)
    cluster_list = new_cluster_list

# Mapeando o resultado final do clustering para os pontos originais
final_clusters = {}
for cluster_id, cluster_indices in enumerate(cluster_list):
    for i in cluster_indices:
        final_clusters[i] = cluster_id

print("quantidade de clusters: ", len(cluster_list))
print(cluster_list)
```

```python
groups = cluster_list

# Preparando dados para o scatterplot
x_coords = [data[i][0] for i in range(len(data))]
y_coords = [data[i][1] for i in range(len(data))]

# Criando um scatterplot
plt.figure(figsize=(10, 8))

# Iterando pelos grupos para plotar os pontos e as circunferências
for idx, group in enumerate(groups):
    x_group = [x_coords[i] for i in group]
    y_group = [y_coords[i] for i in group]

    # Plotar os pontos do grupo com a mesma cor
    plt.scatter(x_group, y_group, label=f'Grupo {idx}', alpha=0.7)

    # Calcular o centro do grupo
    center_x = sum(x_group) / len(x_group)
    center_y = sum(y_group) / len(y_group)

    # Calcular o raio da circunferência do grupo
    max_distance = max([(x - center_x) ** 2 + (y - center_y) ** 2 for x, y in zip(x_group, y_group)])
    circle_radius = max_distance ** 0.5

    # Plotar a circunferência do grupo
    circle = plt.Circle((center_x, center_y), circle_radius, color='black', fill=False)
    plt.gca().add_patch(circle)

# Definir rótulos dos eixos e título do gráfico
plt.xlabel('Coordenada X')
plt.ylabel('Coordenada Y')
plt.title('Scatterplot com Grupos e Circunferências')

# Mostrar legenda
plt.legend()

# Exibir o scatterplot
plt.show()
```

# Gerando o ClusterHead de cada Grupo

```python
cluster_heads = []

rng = np.random.RandomState(seed=seed_value)

topology  = []

for cluster in cluster_list:
    if len(cluster) > 1:
        cluster_head = rng.choice(cluster)
        cluster_heads.append(cluster_head)
        if len(cluster) != 0:
            for node in cluster:
                if node != cluster_head:
                    link = (cluster_head,node)
                    topology.append(link)
```

```python
ch_coords = {}

for key in processing_node:
    ch_coords[key] = data[key]
```

```python
def calcular_angulo_polar(ponto, centroide):
    dx = ponto[0] - centroide[0]
    dy = ponto[1] - centroide[1]
    return math.atan2(dy, dx)

def criar_poligono(pontos):
    if len(pontos) < 3:
        return None

    centroide_x = sum(x for x, y in pontos) / len(pontos)
    centroide_y = sum(y for x, y in pontos) / len(pontos)
    centroide = (centroide_x, centroide_y)

    pontos_ordenados = sorted(pontos, key=lambda ponto: calcular_angulo_polar(ponto, centroide))
    poligono = []

    for i in range(len(pontos_ordenados)):
        ponto_atual = pontos_ordenados[i]
        ponto_proximo = pontos_ordenados[(i + 1) % len(pontos_ordenados)]  # Volte ao primeiro ponto após o último
        poligono.append((ponto_atual, ponto_proximo))

    return poligono

# Converter dicionário em lista de pontos
pontos = list(ch_coords.values())

# Criar o polígono
poligono = criar_poligono(pontos)

# Imprimir as arestas do polígono
for aresta in poligono:
    print(aresta)
```

```python
import matplotlib.pyplot as plt

# ... Seu código para calcular o polígono ...

# Extrair as coordenadas x e y das arestas do polígono
x_coords = [ponto[0] for aresta in poligono for ponto in aresta]
y_coords = [ponto[1] for aresta in poligono for ponto in aresta]

# Adicionar o primeiro ponto para fechar o polígono
x_coords.append(x_coords[0])
y_coords.append(y_coords[0])

# Criar o gráfico
plt.figure(figsize=(8, 6))
plt.plot(x_coords, y_coords, marker='o', linestyle='-', markersize=8, color='b')
plt.xlabel('Coordenada X')
plt.ylabel('Coordenada Y')
plt.title('Polígono Gerado pelos Pontos')
plt.grid(True)
plt.show()
```

```python
# Função para encontrar o vértice correspondente no dicionário
def find_vertex(coord):
    for vertex, point in ch_coords.items():
        if point == coord:
            return vertex

# Criar a nova lista de conexões usando os vértices
nova_lista_conexoes = []
for aresta in poligono:
    v1 = find_vertex(aresta[0])
    v2 = find_vertex(aresta[1])
    nova_lista_conexoes.append((v1, v2))

print(nova_lista_conexoes)
```
