```python
import matplotlib.pyplot as plt
import numpy as np
import math
import random
```

```python
# Constantes
BOLTZMANN = 1.380649e-23  # constante de Boltzmann
temperatura_kelvin = 290
figura_ruido_db = 10.0
largura_banda_hz = 100e6
potencia_transmissao_dbm = 30.0
freq_portadora_hz = 3.5e9
eficiencia_codec = 0.5
snr_minimo_db = 0.0
sigma_shadowing_db = 6.0  # cenário urbano típico

# Dado fixo a ser transmitido (em bits)
dado_bits = 1.5e6    # 10 MB em bits

# Função de perda de caminho com shadowing
def path_loss_5g(distancia_m):
    pl_db = 28.0 + 22 * math.log10(distancia_m) + 20 * math.log10(freq_portadora_hz / 1e9)
    pl_db += random.gauss(0, sigma_shadowing_db)
    return 10 ** (-pl_db / 10)  # ganho linear

# Função para calcular latência de transmissão
def calcular_latencia_5g(distancia_m):
    ganho = path_loss_5g(distancia_m)
    potencia_w = 10 ** (potencia_transmissao_dbm / 10) / 1000
    ruido_w_hz = BOLTZMANN * temperatura_kelvin * (10 ** (figura_ruido_db / 10))
    snr_linear = (ganho * potencia_w) / (ruido_w_hz * largura_banda_hz)
    snr_linear = max(snr_linear, 10 ** (snr_minimo_db / 10))
    taxa_bps = largura_banda_hz * math.log2(1 + snr_linear) * eficiencia_codec
    latencia_ms = (dado_bits / taxa_bps) * 1000
    return latencia_ms

# Geração de dados para o gráfico
distancias = np.linspace(10, 1000, 100)
latencias = [calcular_latencia_5g(d) for d in distancias]

# Plot do gráfico
plt.figure(figsize=(10, 6))
plt.plot(distancias, latencias, label='Latência (ms)', color='blue')
plt.title("Latência de Transmissão 5G vs Distância (Cenário Urbano)")
plt.xlabel("Distância até o gNB (m)")
plt.ylabel("Latência (ms)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Constantes
BOLTZMANN = 1.380649e-23  # constante de Boltzmann
temperatura_kelvin = 290
figura_ruido_db = 10.0
largura_banda_hz = 100e6
potencia_transmissao_dbm = 30.0
freq_portadora_hz = 3.5e9
eficiencia_codec = 0.5
snr_minimo_db = 0.0
sigma_shadowing_db = 6.0  # cenário urbano típico

# Dado fixo a ser transmitido (em bits)
dado_bits = 2.6e6    # 10 MB em bits

# Função de perda de caminho com shadowing
def path_loss_5g(distancia_m):
    pl_db = 28.0 + 22 * math.log10(distancia_m) + 20 * math.log10(freq_portadora_hz / 1e9)
    pl_db += random.gauss(0, sigma_shadowing_db)
    return 10 ** (-pl_db / 10)  # ganho linear

# Função para calcular latência de transmissão
def calcular_latencia_5g(distancia_m):
    ganho = path_loss_5g(distancia_m)
    potencia_w = 10 ** (potencia_transmissao_dbm / 10) / 1000
    ruido_w_hz = BOLTZMANN * temperatura_kelvin * (10 ** (figura_ruido_db / 10))
    snr_linear = (ganho * potencia_w) / (ruido_w_hz * largura_banda_hz)
    snr_linear = max(snr_linear, 10 ** (snr_minimo_db / 10))
    taxa_bps = largura_banda_hz * math.log2(1 + snr_linear) * eficiencia_codec
    latencia_ms = (dado_bits / taxa_bps) * 1000
    return latencia_ms

# Geração de dados para o gráfico
distancias = np.linspace(10, 1000, 100)
latencias = [calcular_latencia_5g(d) for d in distancias]

# Plot do gráfico
plt.figure(figsize=(10, 6))
plt.plot(distancias, latencias, label='Latência (ms)', color='blue')
plt.title("Latência de Transmissão 5G vs Distância (Cenário Urbano)")
plt.xlabel("Distância até o gNB (m)")
plt.ylabel("Latência (ms)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

# Gráfico 3D

```python
import plotly.graph_objs as go

# Geração de grades de distância e tamanho de dados
distancias_3d = np.linspace(10, 1000, 50)              # 10 m a 1000 m
dados_bits_3d = np.linspace(1.5e6, 2.6e6, 50)          # 1.5 Mbits a 2.6 Mbits

# Meshgrid para o gráfico 3D
D, B = np.meshgrid(distancias_3d, dados_bits_3d)

# Calcular latências para cada par (distância, bits)
def latencia(d, b):
    ganho = path_loss_5g(d)
    potencia_w = 10 ** (potencia_transmissao_dbm / 10) / 1000
    ruido_w_hz = BOLTZMANN * temperatura_kelvin * (10 ** (figura_ruido_db / 10))
    snr_linear = (ganho * potencia_w) / (ruido_w_hz * largura_banda_hz)
    snr_linear = max(snr_linear, 10 ** (snr_minimo_db / 10))
    taxa_bps = largura_banda_hz * math.log2(1 + snr_linear) * eficiencia_codec
    return (b / taxa_bps) * 1000  # ms

# Gerar matriz de latência
Z = np.array([[latencia(d, b) for d in distancias_3d] for b in dados_bits_3d])

# Criar gráfico 3D interativo
fig = go.Figure(data=[go.Surface(z=Z, x=distancias_3d, y=dados_bits_3d, colorscale='Viridis')])
fig.update_layout(
    title='Latência 5G vs Distância e Tamanho de Dados (Cenário Urbano)',
    scene=dict(
        xaxis_title='Distância (m)',
        yaxis_title='Tamanho dos Dados (bits)',
        zaxis_title='Latência (ms)'
    ),
    autosize=True,
    margin=dict(l=0, r=0, b=0, t=40)
)
fig.show()
```
